# 04 — Reward model

El reward model es un **juez de calidad**, no el generador de actividades. Primero se ejecuta una prueba de humo y luego, solo si el corpus tiene señal suficiente, el entrenamiento real. No se promueve con Spearman menor que 0.5.

```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RUN_REAL_REWARD_TRAINING = False
command = [sys.executable, str(ROOT / "scripts/aulatex_training/train_reward_model.py")]
if RUN_REAL_REWARD_TRAINING:
    command += ["--train-file", str(ROOT / "data/private/processed/reward.jsonl"), "--output-dir", str(ROOT / "models/local/reward-model")]
else:
    command += ["--smoke-test", "--output-dir", str(ROOT / "models/local/reward-smoke")]
print("Ejecutando:", command)
result = subprocess.run(command, cwd=ROOT, text=True)
print("Código de salida:", result.returncode)
```
