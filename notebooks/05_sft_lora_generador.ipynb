# 05 — SFT LoRA/QLoRA del generador

Ajusta un modelo instruct mediante un adaptador LoRA/QLoRA en la A10G. Por seguridad empieza en `dry-run`; el entrenamiento real requiere cambiar explícitamente `RUN_TRAINING = True`. Los pesos base no se duplican y el adaptador queda fuera de Git.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RUN_TRAINING = False
SMOKE_ONLY = True
command = [
    sys.executable, "-m", "scripts.aulatex_training.train_generator",
    "--config", str(ROOT / "scripts/aulatex_training/configs/sft-a10g.json"),
]
if RUN_TRAINING:
    command.append("--run")
    if SMOKE_ONLY:
        command.append("--smoke")
print("Modo:", "ENTRENAMIENTO" if RUN_TRAINING else "DRY-RUN")
result = subprocess.run(command, cwd=ROOT, text=True)
print("Código de salida:", result.returncode)
```
